```python
import shap
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the best model (Gradient Boosting)
model = joblib.load('../models/best_model.pkl')

# Load the test data
X_test = pd.read_csv('../data/processed/X_test.csv')
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()
```

```python
# Build a TreeExplainer for fast, exact SHAP values
explainer = shap.TreeExplainer(model)

# Compute SHAP values on the test set
shap_values = explainer(X_test)
```

```python
# Generate the SHAP beeswarm summary plot
shap.plots.beeswarm(shap_values, show=False)
plt.tight_layout()
plt.savefig('../reports/figures/shap_summary.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Calculate mean absolute SHAP value per feature
mean_abs_shap = np.abs(shap_values.values).mean(axis=0)
feature_importance = pd.DataFrame({
    'feature': X_test.columns,
    'importance': mean_abs_shap
}).sort_values('importance', ascending=False)

# Take top 5 features
top_5 = feature_importance.head(5).copy()

# Map technical feature names to business-friendly labels
label_map = {
    'Contract_Two year': 'Two-year contract',
    'Contract_One year': 'One-year contract',
    'tenure': 'Tenure (months as customer)',
    'InternetService_Fiber optic': 'Fiber optic internet',
    'PaperlessBilling_Yes': 'Paperless billing',
}
top_5['label'] = top_5['feature'].map(label_map).fillna(top_5['feature'])
```

```python
# Create the Churn Risk Scorecard
fig, ax = plt.subplots(figsize=(8, 4))
ax.barh(top_5['label'], top_5['importance'], color='#e74c3c')
ax.set_xlabel('Mean |SHAP value| (impact on churn prediction)')
ax.set_title('Churn Risk Scorecard: Top 5 Drivers')
ax.invert_yaxis()  # Highest importance at the top
plt.tight_layout()
plt.savefig('../reports/figures/churn_risk_scorecard.png', dpi=150, bbox_inches='tight')
plt.show()
```
